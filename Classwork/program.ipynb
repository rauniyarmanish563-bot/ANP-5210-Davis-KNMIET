```python
import numpy as np

num = int(input("Enter a number: "))

if num == 0:
    print("Zero has infinitely many factors.")
else:
    n = abs(num)
    even_factors = []

    for i in range(1, n + 1):
        if n % i == 0 and i % 2 == 0:
            even_factors.append(i)
            even_factors.append(-i)

    arr = np.array(even_factors)
    print("Even factors:", even_factors)
    print("NumPy array:", arr)
```
